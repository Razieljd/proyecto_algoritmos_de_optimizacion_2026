```python
mejor_coste = float("inf")
mejor_solucion = None

def calcular_coste(dias, actores_por_toma):
    coste = 0
    for dia in dias:
        actores_dia = set()
        for toma in dia:
            actores_dia |= actores_por_toma[toma]
        coste += len(actores_dia)
    return coste


def backtracking(toma_actual, dias, actores_por_toma, total_tomas):
    global mejor_coste, mejor_solucion

    if toma_actual == total_tomas:
        coste = calcular_coste(dias, actores_por_toma)
        if coste < mejor_coste:
            mejor_coste = coste
            mejor_solucion = [d.copy() for d in dias]
        return

    # Intentar en días existentes
    for dia in dias:
        if len(dia) < 6:
            dia.append(toma_actual)
            backtracking(toma_actual + 1, dias, actores_por_toma, total_tomas)
            dia.pop()

    # Crear nuevo día
    dias.append([toma_actual])
    backtracking(toma_actual + 1, dias, actores_por_toma, total_tomas)
    dias.pop()
```

```python
import random

def construir_solucion(actores_por_toma, alpha=0.3):

    dias = []
    actores_por_dia = []

    tomas = list(range(len(actores_por_toma)))
    random.shuffle(tomas)

    for toma in tomas:

        candidatos = []

        for i in range(len(dias)):
            if len(dias[i]) < 6:
                incremento = len(actores_por_toma[toma] - actores_por_dia[i])
                candidatos.append((i, incremento))

        # Si no hay día posible, creamos uno nuevo
        if candidatos == []:
            dias.append([toma])
            actores_por_dia.append(set(actores_por_toma[toma]))
            continue

        # Construir la Lista Restringida de Candidatos (LRC)
        incrementos = [c[1] for c in candidatos]
        incremento_minimo = min(incrementos)
        incremento_maximo = max(incrementos)

        limite = incremento_minimo + alpha * (incremento_maximo - incremento_minimo)

        LRC = [c for c in candidatos if c[1] <= limite]

        elegido = random.choice(LRC)
        dia_elegido = elegido[0]

        dias[dia_elegido].append(toma)
        actores_por_dia[dia_elegido].update(actores_por_toma[toma])

    return dias
```

```python
def calcular_coste(dias, actores_por_toma):
    coste = 0
    for dia in dias:
        actores_dia = set()
        for toma in dia:
            for actor in actores_por_toma[toma]:
                actores_dia.add(actor)
        coste += len(actores_dia)
    return coste

def busqueda_local(dias, actores_por_toma):
    mejora = True
    while mejora:
        mejora = False
        coste_actual = calcular_coste(dias, actores_por_toma)
        
        for i in range(len(dias)):
            for toma in dias[i][:]:

                for j in range(len(dias)):
                    if i != j and len(dias[j]) < 6:

                        # probar mover
                        dias[i].remove(toma)
                        dias[j].append(toma)

                        nuevo_coste = calcular_coste(dias, actores_por_toma)

                        if nuevo_coste < coste_actual:
                            mejora = True
                            coste_actual = nuevo_coste
                            break
                        else:
                            # deshacer
                            dias[j].remove(toma)
                            dias[i].append(toma)
                if mejora:
                    break
            if mejora:
                break
    return dias
```

```python
def GRASP(actores_por_toma, iteraciones=50, alpha=0.3):

    mejor_solucion = None
    mejor_coste = float("inf")

    for _ in range(iteraciones):

        solucion = construir_solucion(actores_por_toma, alpha)
        solucion = busqueda_local(solucion, actores_por_toma)

        coste = calcular_coste(solucion, actores_por_toma)

        if coste < mejor_coste:
            mejor_coste = coste
            mejor_solucion = solucion

    return mejor_solucion, mejor_coste
```

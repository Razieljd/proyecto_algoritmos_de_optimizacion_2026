Respuesta

**Posibilidades sin restricciones:**
no tenemos ninguna restricción por lo tanto tenemos 30 tomas, sin repetición y nos importa el orden se usaría una permutación  sin repetición:

$$P=n!$$
El número de posibilidades sin restricciones viene dado por : $30!$

**Posibilidades con restricciones:**

obtenemos el minimo número de días $30/6=5$
utilizando esta formula
$$C_{n,r}=\frac{n!}{r!(n-r)!}$$
día 1 
$$C_{30,6}=\frac{30!}{6!(24)!}$$
día 2
$$C_{24,6}=\frac{24!}{6!(18)!}$$
día 3
$$C_{18,6}=\frac{18!}{6!(12)!}$$
día 4
$$C_{12,6}=\frac{12!}{6!(6)!}$$
día 5
$$C_{6,6}=\frac{6!}{6!(0)!}=1$$

por teorema de bayes tenemos:
$$\frac{30!}{6!(24)!} * \frac{24!}{6!(18)!} * \frac{18!}{6!(12)!} * \frac{12!}{6!(6)!} * 1 = \frac{30!}{(6!)^5}$$

Respuesta

```python
def optimizar_rodaje(lista_actores_tomas, tomas_por_dia=6):
    """
    Función maestra de Branch and Bound + DP con cupo elástico.
    """
    
    # --- 1. PREPARACIÓN (BITMASKING) ---
    tomas_bits = []
    for actores in lista_actores_tomas:
        mask = 0
        for a in actores: mask |= (1 << a)
        tomas_bits.append(mask)

    num_tomas = len(tomas_bits)
    # AJUSTE 1: Redondeo hacia arriba para cubrir todas las tomas
    num_dias = math.ceil(num_tomas / tomas_por_dia)

    mejor_costo = float('inf')
    mejor_plan = []
    memo = {}
    inicio_t = time.time()

    print(f"\n INICIANDO: ({num_tomas} tomas, {num_dias} días)")

    # --- 2. LÓGICA DE PODA (LOWER BOUND) ---
    def obtener_cota(mask_tomas, costo_actual):
        actores_pendientes = 0
        for i in range(num_tomas):
            if not (mask_tomas & (1 << i)):
                actores_pendientes |= tomas_bits[i]
        return costo_actual + bin(actores_pendientes).count('1')

    # --- 3. RECURSIÓN (BRANCH AND BOUND) ---
    def backtrack(mask_tomas, costo_acum, plan, dia_actual):
        nonlocal mejor_costo, mejor_plan
        
        # AJUSTE 2: El caso base ahora es cuando la máscara está llena (todas las tomas hechas)
        if mask_tomas == (1 << num_tomas) - 1:
            if costo_acum < mejor_costo:
                mejor_costo = costo_acum
                mejor_plan = plan
                print(f" Nuevo óptimo parcial: {mejor_costo}")
            return

        if obtener_cota(mask_tomas, costo_acum) >= mejor_costo:
            return

        if mask_tomas in memo and memo[mask_tomas] <= costo_acum:
            return
        memo[mask_tomas] = costo_acum

        # Ramificación
        faltan = [i for i in range(num_tomas) if not (mask_tomas & (1 << i))]
        if not faltan: return # Seguridad
        
        semilla = faltan[0]
        # AJUSTE 3: El cupo de hoy es 6, o lo que quede si es el último día
        cupo_hoy = min(tomas_por_dia, len(faltan))
        
        # Candidatas: El resto de las tomas que faltan (limitamos a 12 para velocidad)
        resto_faltan = faltan[1:]
        candidatas = resto_faltan[:12] 

        def generar_dia(idx, actual):
            # El día se completa cuando tenemos (cupo_hoy - 1) compañeras para la semilla
            if len(actual) == cupo_hoy - 1:
                grupo = [semilla] + actual
                m_dia_tomas = mask_tomas
                m_actores_dia = 0
                for t in grupo:
                    m_dia_tomas |= (1 << t)
                    m_actores_dia |= tomas_bits[t]
                
                backtrack(m_dia_tomas, costo_acum + bin(m_actores_dia).count('1'), 
                          plan + [grupo], dia_actual + 1)
                return

            for i in range(idx, len(candidatas)):
                actual.append(candidatas[i])
                generar_dia(i + 1, actual)
                actual.pop()

        generar_dia(0, [])

    # Ejecución
    backtrack(0, 0, [], 0)
    
    # --- 4. RESULTADOS ---
    fin_t = time.time()
    print(f"Costo Final: {mejor_costo} |{mejor_plan}| Tiempo: {fin_t - inicio_t:.2f}s")
    return mejor_costo, mejor_plan
    
costo_A, plan_A = optimizar_rodaje(actores_por_toma)
```

La complejidad es exponencial debido a la naturaleza combinatoria del problema. Sin embargo, mediante el uso de Programación Dinámica se reduce la redundancia de $O(N!)$ a $O(2^N)$, y mediante la Poda de Branch and Bound, el número de nodos realmente visitados se mantiene dentro de límites computacionalmente tratables para $N=30$, permitiendo encontrar el óptimo en tiempo polinomial para este caso específico.

```python
import random

# m (tomas) > 6: Evita que el rodaje sea demasiado corto o simple.
# n (actores) > 2: Garantiza complejidad en la agenda.
# Regla: No puede haber tomas vacías; cada fila debe tener al menos un 1 (actor).
m= 11
n= 5

def generador_datos_entrada(m, n):
    matriz = []
    for _ in range(m):
        fila = [random.randint(0, 1) for _ in range(n)]
        # Si la fila es todo ceros, elegimos un actor al azar para que aparezca
        if sum(fila) == 0:
            fila[random.randint(0, n-1)] = 1
        matriz.append(fila)
    return matriz

matriz_tomas = generador_datos_entrada(m,n)
for fila in matriz_tomas:
    print(fila)
```

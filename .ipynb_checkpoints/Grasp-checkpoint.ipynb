```python
import random
```

```python
import urllib.request
import csv
url = "https://docs.google.com/spreadsheets/d/1Ipn6IrbQP4ax8zOnivdBIw2lN0JISkJG4fXndYd27U0/export?format=csv&gid=0"
try:
    print("Descargando datos...")
    respuesta = urllib.request.urlopen(url)
    lineas = [linea.decode('utf-8') for linea in respuesta.readlines()]
    lineas = lineas[1:] ## eliminamos la primera fila que no la necesitamos
    lineas.pop(-2)
    totales_actores = lineas.pop(-1)
    print(totales_actores)
    lector_csv = csv.reader(lineas)
    
    encabezados = next(lector_csv)
    datos = [fila for fila in lector_csv]
    
    print(f"Las columnas son: {encabezados}")
    num_tomas = len(datos)
    print(f"Número de tomas : {num_tomas}") ## elimino linea totales
    num_actores = len(datos[0])-3
    print(f"Número de actores : {num_actores}") ## elimino las columnas totales, vacía y toma
    display(datos)

except urllib.error.URLError as e:
    print(f"Error de conexión: {e}")
except Exception as e:
    print(f"Ocurrió un error inesperado: {e}")
```

```python
matriz_tomas = []
list_totales_actores = totales_actores.split(',')
list_totales_actores = list_totales_actores[1:-2]
print(list_totales_actores)
total_tomas_dia = {}
for i in range(len(datos)):
    matriz_tomas.append([int(x) for x in datos[i][1:-2]])
    for j in range(len(datos[0])):
        if j == len(datos[0])-1:
            total_tomas_dia[int(i+1)]=int(datos[i][-1])
display(matriz_tomas)
print("total de tomas en el día")
print(total_tomas_dia)
```

```python
# usamos los datos de la matriz para construir el conjunto de actores por toma
actores_por_toma = []

for fila in matriz_tomas:
    actores = set()
    for indice_actor in range(len(fila)):
        if fila[indice_actor] == 1:
            actores.add(indice_actor)
    actores_por_toma.append(actores)

print("Actores por toma:")
for i, a in enumerate(actores_por_toma, start=1):
    print(f"Toma {i}: {a}")
```

```python
def construir_solucion(actores_por_toma, alpha=0.3):

    dias = []
    actores_por_dia = []

    tomas = list(range(len(actores_por_toma)))
    random.shuffle(tomas)

    for toma in tomas:

        candidatos = []

        for i in range(len(dias)):
            if len(dias[i]) < 6:
                incremento = len(actores_por_toma[toma] - actores_por_dia[i])
                candidatos.append((i, incremento))

        # si no hay candidatos posibles. Se crea un nuevo día
        if candidatos == []:
            dias.append([toma])
            actores_por_dia.append(set(actores_por_toma[toma]))
            continue

        incrementos = [c[1] for c in candidatos]
        incremento_minimo = min(incrementos)
        incremento_maximo = max(incrementos)

        limite = incremento_minimo + alpha * (incremento_maximo - incremento_minimo)

        LRC = [c for c in candidatos if c[1] <= limite]

        elegido = random.choice(LRC)
        dia_elegido = elegido[0]

        dias[dia_elegido].append(toma)
        actores_por_dia[dia_elegido].update(actores_por_toma[toma])

    print("\nSolución construida:", dias)
    return dias
```

```python
def calcular_coste(dias, actores_por_toma):
    coste = 0
    for dia in dias:
        actores_dia = set()
        for toma in dia:
            for actor in actores_por_toma[toma]:
                actores_dia.add(actor)
        coste += len(actores_dia)
    return coste
    
def busqueda_local(dias, actores_por_toma):

    mejora = True
    iteracion = 0
    while mejora:
        mejora = False
        iteracion += 1
        print("Iteración búsqueda local:", iteracion)
        coste_actual = calcular_coste(dias, actores_por_toma)

        print("\nBúsqueda local - coste actual:", coste_actual)

        for i in range(len(dias)):
            for toma in dias[i][:]:
                for j in range(len(dias)):
                    if i != j and len(dias[j]) < 6:

                        dias[i].remove(toma)
                        dias[j].append(toma)

                        nuevo_coste = calcular_coste(dias, actores_por_toma)

                        if nuevo_coste < coste_actual:
                            print("Mejora encontrada")
                            mejora = True
                            coste_actual = nuevo_coste
                            break
                        else:
                            dias[j].remove(toma)
                            dias[i].append(toma)

                if mejora:
                    break
            if mejora:
                break

    return dias
```

```python
def GRASP(actores_por_toma, iteraciones=10, alpha=0.3):

    mejor_solucion = None
    mejor_coste = float("inf")

    for i in range(iteraciones):

        print(f"\n========== ITERACIÓN {i+1} ==========")

        solucion = construir_solucion(actores_por_toma, alpha)
        solucion = busqueda_local(solucion, actores_por_toma)

        coste = calcular_coste(solucion, actores_por_toma)

        print("Coste obtenido:", coste)

        if coste < mejor_coste:
            print("Nueva mejor solución encontrada")
            mejor_coste = coste
            mejor_solucion = solucion

    print("\n====== RESULTADO FINAL ======")
    print("Mejor coste:", mejor_coste)
    print("Mejor solucion:", mejor_solucion)

    return mejor_solucion, mejor_coste
```

```python
GRASP(actores_por_toma)
```
